```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32


# Ask the user for a temperature
celsius = float(input("Enter a temperature in °C: "))

# Convert the temperature to Fahrenheit
fahrenheit = celsius_to_fahrenheit(celsius)

# Display the converted temperature
print(f"{celsius}°C = {fahrenheit:.2f}°F")


# Check whether the temperature is above or below a threshold
threshold = 30

if celsius > threshold:
    print(f"The temperature is above {threshold}°C.")
elif celsius < threshold:
    print(f"The temperature is below {threshold}°C.")
else:
    print(f"The temperature is exactly {threshold}°C.")


# Use a loop to convert several temperatures
print("\nConverting several temperatures:")

temperatures = [0, 10, 20, 30, 40]

for temp in temperatures:
    converted = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {converted:.2f}°F")
```
